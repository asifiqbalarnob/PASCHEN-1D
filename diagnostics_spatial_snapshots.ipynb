# PASCHEN-1D Instantaneous Spatial Diagnostics

This notebook plots saved spatial profiles at selected snapshot times. Each diagnostic has one definition cell and one independent plotting cell.

Figures are not saved unless `save = True` in the corresponding plotting cell.


## Diagnostic Index

**Spatial profile diagnostics:** `ne`, `ni`, `phi`, `E`, `Gamma_i`, `Gamma_e`, `townsend_alpha`, `nu_i`, `S_ion`, `S`, `mu_e`, `D_e`.

**Sheath spatial-context diagnostics:** `edge_x_m`, `width_m`, `voltage_drop_abs_V`, `peak_abs_E_V_m`, `space_charge_sigma_C_m2`, `space_charge_Q_C`.

**Grouped plot:** edit `custom_group = (...)` in the final cell.


```python
from pathlib import Path

import matplotlib.pyplot as plt

from diagnostics_io import (
    available_spatial,
    available_temporal,
    compute_sheath_rows,
    discover_runs,
    load_run_context,
    plot_averaged_spatial_group,
    plot_averaged_spatial_profile,
    plot_sheath_spatial_context,
    plot_sheath_temporal_metric,
    plot_spatial_group_snapshot,
    plot_spatial_profiles,
    plot_spatial_temporal_summary,
    plot_temporal_group,
    plot_temporal_quantity,
    print_available_diagnostics,
    print_run_summary,
    print_sheath_validity,
    set_notebook_plot_style,
)

set_notebook_plot_style(font_size=11, figure_dpi=150, savefig_dpi=600)

project_dir = Path('.')
available_runs = discover_runs(project_dir)
print('Saved runs found:')
for name in available_runs:
    print('  -', name)

# Select the run folder to postprocess. If this example run is not present,
# set run_name manually to one of the folders printed above.
run_name = 'argon_photoemission_induced_discharge'
# run_name = 'your_saved_run_folder_name'

if run_name not in available_runs:
    if len(available_runs) == 1:
        run_name = available_runs[0]
    else:
        raise RuntimeError('Set run_name to one of the saved run folders printed above.')

ctx = load_run_context(run_name, project_dir)
print_run_summary(ctx)
print_available_diagnostics(ctx)
```

## Spatial Profiles

The default `t_samples` list samples profiles from 1 to 10 microseconds. Set `t_samples = None` to plot only the final saved snapshot.

**Definition.** Electron number density, $n_e(x,t)$.


```python
# Plot ne at selected snapshot times.
quantity = 'ne'

# Figure saving controls
save = True
fig_name = f'{quantity}_spatial_snapshots'

# Snapshot controls
# t_samples = None          # None -> final snapshot; or tuple in seconds, e.g. (1e-6, 2e-6, 5e-6)
t_samples = (1e-6, 1.5e-6, 2e-6, 2.5e-6, 3e-6, 3.5e-6, 4e-6, 4.5e-6, 5e-6, 5.5e-6, 6e-6, 6.5e-6, 7e-6, 7.5e-6, 8e-6, 8.5e-6, 9e-6, 9.5e-6, 10e-6)
# t_samples = (80e-9, 86e-9, 88e-9, 90e-9)
             
# Plot controls
x_unit = 'cm'
value_mode = 'raw'          # 'raw', 'abs', 'positive', or 'negative_abs'
y_scale_factor = 1.0
ylabel = r'$n_e$ [m$^{-3}$]'
title = f'{quantity} profiles'
xscale = 'linear'
yscale = 'log'
figsize = (5.0, 3.5)
cmap = 'viridis'
linewidth = 2.0
linestyle = '-'
grid = True
legend = True
legend_loc = 'best'
colorbar = True
time_unit_for_labels = 'us'
xlim = None
ylim = None

if quantity not in available_spatial(ctx):
    print(f'Skipping {quantity}: this run did not save that spatial diagnostic.')
else:
    plot_spatial_profiles(
        ctx,
        quantity,
        t_samples=t_samples,
        x_unit=x_unit,
        value_mode=value_mode,
        y_scale_factor=y_scale_factor,
        ylabel=ylabel,
        title=title,
        xscale=xscale,
        yscale=yscale,
        figsize=figsize,
        cmap=cmap,
        linewidth=linewidth,
        linestyle=linestyle,
        grid=grid,
        legend=legend,
        legend_loc=legend_loc,
        colorbar=colorbar,
        time_unit_for_labels=time_unit_for_labels,
        xlim=xlim,
        ylim=ylim,
        save=save,
        fig_name=fig_name,
    )
```
