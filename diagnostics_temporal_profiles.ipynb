# PASCHEN-1D Temporal Diagnostics

This notebook plots scalar time histories and time-dependent summaries derived from saved PASCHEN-1D outputs. Each diagnostic has a short definition cell followed by one independent plotting cell. Edit only the cells you need.

Figures are not saved unless `save = True` in the corresponding plotting cell.


## Diagnostic Index

**Scalar temporal diagnostics:** `V_app`, `V_gap`, `V_node`, `V_source`, `I_discharge`, `I_transport_plasma`, `I_transport_circuit`, `I_emission_circuit`, `I_emission_area`, `I_displacement_gap`, `cfl`, `diffusion_cfl`, `picard_iterations`, `adaptive_substeps`, `adaptive_dt_sub`, `adaptive_cfl_est`, `adaptive_diffusion_cfl_est`.

**Spatial diagnostics summarized vs time:** `ne`, `ni`, `phi`, `E`, `Gamma_i`, `Gamma_e`, `townsend_alpha`, `nu_i`, `S_ion`, `S`, `mu_e`, `D_e`.

**Sheath temporal diagnostics:** `edge_x_m`, `width_m`, `voltage_drop_abs_V`, `peak_abs_E_V_m`, `space_charge_sigma_C_m2`, `space_charge_Q_C`.

**Grouped plot:** edit `custom_group = (...)` in the final cell.


```python
from pathlib import Path

import matplotlib.pyplot as plt

from diagnostics_io import (
    available_spatial,
    available_temporal,
    compute_sheath_rows,
    discover_runs,
    load_run_context,
    plot_averaged_spatial_group,
    plot_averaged_spatial_profile,
    plot_sheath_spatial_context,
    plot_sheath_temporal_metric,
    plot_spatial_group_snapshot,
    plot_spatial_profiles,
    plot_spatial_temporal_summary,
    plot_temporal_group,
    plot_temporal_quantity,
    print_available_diagnostics,
    print_run_summary,
    print_sheath_validity,
    set_notebook_plot_style,
)

set_notebook_plot_style(font_size=11, figure_dpi=150, savefig_dpi=600)

project_dir = Path('.')
available_runs = discover_runs(project_dir)
print('Saved runs found:')
for name in available_runs:
    print('  -', name)

# Select the run folder to postprocess. If this example run is not present,
# set run_name manually to one of the folders printed above.
run_name = 'argon_photoemission_induced_discharge'
# run_name = 'argon_dc_glow_discharge_140V'
# run_name = 'nitrogen_pulsed_discharge'

if run_name not in available_runs:
    if len(available_runs) == 1:
        run_name = available_runs[0]
    else:
        raise RuntimeError('Set run_name to one of the saved run folders printed above.')

ctx = load_run_context(run_name, project_dir)
print_run_summary(ctx)
print_available_diagnostics(ctx)
```

### Temporal: `V_app`

**Definition.** Applied source voltage reconstructed from waveform metadata. For a DC case, $V_\mathrm{app}(t)=V_\mathrm{peak}$; for pulsed/RF cases, it follows the waveform saved in `run_metadata.json`.


```python
# Plot V_app as an independent scalar time history.
quantity = 'V_app'

# Figure saving controls
save = False
fig_name = f'{quantity}_temporal'

# Plot controls
t_start = None
t_end = None
t_unit = 'us'
value_mode = 'raw'          # 'raw', 'abs', 'positive', or 'negative_abs'
y_scale_factor = 1.0
ylabel = 'Voltage [V]'
title = quantity
xscale = 'linear'
yscale = 'linear'
figsize = (5.0, 3.5)
color = 'C7'
linestyle = '-'
linewidth = 2.0
marker = None
grid = True
legend = True
legend_loc = 'best'
xlim = None
ylim = None
max_points = 30000

if quantity not in available_temporal(ctx):
    print(f'Skipping {quantity}: this run did not save that diagnostic.')
else:
    plot_temporal_quantity(
        ctx,
        quantity,
        t_start=t_start,
        t_end=t_end,
        t_unit=t_unit,
        value_mode=value_mode,
        y_scale_factor=y_scale_factor,
        ylabel=ylabel,
        title=title,
        xscale=xscale,
        yscale=yscale,
        figsize=figsize,
        color=color,
        linestyle=linestyle,
        linewidth=linewidth,
        marker=marker,
        grid=grid,
        legend=legend,
        legend_loc=legend_loc,
        xlim=xlim,
        ylim=ylim,
        max_points=max_points,
        save=save,
        fig_name=fig_name,
    )
```

## Sheath Temporal Diagnostics

The sheath edge is inferred from a quasineutrality metric,

$$\eta(x,t)=\frac{|n_i(x,t)-n_e(x,t)|}{\max\left[\frac{1}{2}(n_i(x,t)+n_e(x,t)),\ n_\mathrm{floor}(t)\right]}.$$

A point is treated as quasineutral when $\eta \le \eta_\mathrm{tol}$ and the local mean density is above the floor. The anode sheath is measured from the anode to the first quasineutral point; the cathode sheath is measured from the cathode to the last quasineutral point.
