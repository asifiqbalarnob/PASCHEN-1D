# PASCHEN-1D Time- and Cycle-Averaged Spatial Diagnostics

This notebook plots averaged spatial profiles from saved PASCHEN-1D snapshots. Averaging can be done over a user-selected time window or over the last `N_cycle_avg` RF cycles when the run metadata contains `f_rf`.

Figures are not saved unless `save = True` in the corresponding plotting cell.


## Diagnostic Index

**Averaged spatial diagnostics:** `ne`, `ni`, `phi`, `E`, `Gamma_i`, `Gamma_e`, `townsend_alpha`, `nu_i`, `S_ion`, `S`, `mu_e`, `D_e`.

**Grouped plot:** edit `custom_group = (...)` in the final cell.


```python
from pathlib import Path

import matplotlib.pyplot as plt

from diagnostics_io import (
    available_spatial,
    available_temporal,
    compute_sheath_rows,
    discover_runs,
    load_run_context,
    plot_averaged_spatial_group,
    plot_averaged_spatial_profile,
    plot_sheath_spatial_context,
    plot_sheath_temporal_metric,
    plot_spatial_group_snapshot,
    plot_spatial_profiles,
    plot_spatial_temporal_summary,
    plot_temporal_group,
    plot_temporal_quantity,
    print_available_diagnostics,
    print_run_summary,
    print_sheath_validity,
    set_notebook_plot_style,
)

set_notebook_plot_style(font_size=11, figure_dpi=150, savefig_dpi=600)

project_dir = Path('.')
available_runs = discover_runs(project_dir)
print('Saved runs found:')
for name in available_runs:
    print('  -', name)

# Select the run folder to postprocess. If this example run is not present,
# set run_name manually to one of the folders printed above.
run_name = 'argon_photoemission_induced_discharge'
# run_name = 'your_saved_run_folder_name'

if run_name not in available_runs:
    if len(available_runs) == 1:
        run_name = available_runs[0]
    else:
        raise RuntimeError('Set run_name to one of the saved run folders printed above.')

ctx = load_run_context(run_name, project_dir)
print_run_summary(ctx)
print_available_diagnostics(ctx)
```

## Averaging Controls

Each cell below has its own averaging controls. Use `averaged_mode = 'time_window'` for direct start/end times or `averaged_mode = 'last_n_cycles'` for RF cycle averaging.


### Averaged Spatial Profile: `ne`

**Definition.** Electron number density, $n_e(x,t)$.


```python
# Plot an averaged spatial profile for ne.
quantity = 'ne'

# Figure saving controls
save = False
fig_name = f'{quantity}_averaged_spatial'

# Averaging controls
averaged_mode = 'time_window'   # 'time_window' or 'last_n_cycles'
t_avg_start = 1e-6
t_avg_end = 10e-6
N_cycle_avg = 1                 # used only when averaged_mode = 'last_n_cycles'
statistic = 'mean'      # 'mean', 'mean_abs', 'rms', 'max', or 'min'

# Plot controls
x_unit = 'cm'
value_mode = 'raw'              # 'raw', 'abs', 'positive', or 'negative_abs'
y_scale_factor = 1.0
ylabel = r'$n_e$ [m$^{-3}$]'
title = f'{quantity} averaged profile'
xscale = 'linear'
yscale = 'log'
figsize = (5.0, 3.5)
color = 'C0'
linestyle = '-'
linewidth = 2.0
grid = True
legend = True
legend_loc = 'best'
xlim = None
ylim = None

if quantity not in available_spatial(ctx):
    print(f'Skipping {quantity}: this run did not save that spatial diagnostic.')
else:
    plot_averaged_spatial_profile(
        ctx,
        quantity,
        averaged_mode=averaged_mode,
        t_avg_start=t_avg_start,
        t_avg_end=t_avg_end,
        N_cycle_avg=N_cycle_avg,
        statistic=statistic,
        value_mode=value_mode,
        x_unit=x_unit,
        y_scale_factor=y_scale_factor,
        ylabel=ylabel,
        title=title,
        xscale=xscale,
        yscale=yscale,
        figsize=figsize,
        color=color,
        linestyle=linestyle,
        linewidth=linewidth,
        grid=grid,
        legend=legend,
        legend_loc=legend_loc,
        xlim=xlim,
        ylim=ylim,
        save=save,
        fig_name=fig_name,
    )
```
